### 0.2 Data Cleaning

```python
# 1. Fill missing embarked with the most frequent value
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
```

```python
print(df['embarked'].isnull().sum())
```

```python
# 2. Convert sex and class to categorical type
df['sex'] = df['sex'].astype('category')
df['class'] = df['class'].astype('category')

```

```python
print(df[['sex', 'class']].dtypes)
```

```python
# 3. Create family_size = sibsp + parch + 1
df['family_size'] = df['sibsp'] + df['parch'] + 1

```

```python
df[['sibsp', 'parch', 'family_size']].head()
```

```python
# 4. Cap fare at the 99th percentile
cap_value = df['fare'].quantile(0.99)
df['fare'] = np.where(df['fare'] > cap_value, cap_value, df['fare'])
```

```python
print("Cap value:", cap_value)
print("Max fare after capping:", df['fare'].max())
```
