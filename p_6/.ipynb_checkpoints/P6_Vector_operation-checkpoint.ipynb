## Q1. Create a Python list representing vector **v = [2, -1, 3]** and multiply it by scalar **4**.

1. Show the output.
2. Explain why the result is incorrect from a linear algebra perspective.
3. Convert the vector to a NumPy array and repeat the operation.

### Answer 1

````python
# 1. Python list times a scalar
v = [2, -1, 3]
result_list = v * 4
print("List result:  ", result_list)

# 3. NumPy array times a scalar
v_np = np.array(v)
result_np = v_np * 4
print("NumPy result: ", result_np)
````

**Explanation:** For a Python list, `*` means *repetition* (concatenation of the list with itself 4 times), so the result is a list of length 12 instead of a vector of length 3. In linear algebra, scalar multiplication must scale **every component** of the vector: 4·[2, -1, 3] = [8, -4, 12]. A NumPy array overloads `*` as element-wise multiplication, which gives the mathematically correct result.

## Q2. Given vector **a = [5, -3]** and scalar **$\lambda$ = -2**:

1. Compute λa.
2. Compare their magnitudes
3. Explain the effect on magnitude and direction.

Hint: to write $\lambda$ in markdown syntax, do the following:

```pyhton
$\lambda $
```

#### Answer 2

````python
a = np.array([5, -3])
lam = -2

# 1. Scalar multiplication
lam_a = lam * a
print("a        =", a)
print("lambda*a =", lam_a)

# 2. Magnitudes
norm_a = np.linalg.norm(a)
norm_lam_a = np.linalg.norm(lam_a)
print(f"|a|        = {norm_a:.4f}")
print(f"|lambda*a| = {norm_lam_a:.4f}")
print(f"ratio |lambda*a| / |a| = {norm_lam_a / norm_a:.4f}")
````

**Explanation:** $\lambda a = [-10, 6]$. The magnitude is scaled by $|\lambda| = 2$: $|\lambda a| = 2|a|$ (≈ 11.66 vs ≈ 5.83). Because $\lambda$ is **negative**, the direction is **reversed** (flipped by 180°), but the new vector stays on the same line through the origin as $a$ (they are parallel/collinear).

## Q3. Create:

- Column vector shape (3,1)
- Row vector shape (1,4)
- Print the shape of the results

Add them using NumPy broadcasting.

### Answer 3

````python
col = np.array([[1], [2], [3]])      # shape (3, 1)
row = np.array([[10, 20, 30, 40]])   # shape (1, 4)

print("Column vector shape:", col.shape)
print("Row vector shape:   ", row.shape)

result = col + row                   # broadcasting -> (3, 4)
print("Result shape:       ", result.shape)
print(result)
````

**Explanation:** Broadcasting compares shapes from the right: (3,1) and (1,4). Dimensions of size 1 are stretched to match the other array, so the column is repeated across 4 columns and the row is repeated down 3 rows. Both become (3,4) and are added element-wise, giving a (3,4) matrix.

## Q4 . Create vector **v = [1,2,3,4]** and add scalar **5**.

Explain why NumPy allows this.

### Answer 4

````python
v = np.array([1, 2, 3, 4])
print(v + 5)
````

**Explanation:** Mathematically, adding a scalar to a vector is not defined, but NumPy uses *broadcasting*: the scalar (shape `()`) is stretched to the shape of the array and added to every element, so `v + 5` = `[6, 7, 8, 9]`.

## Q5. Given vector **a = [1,2,3,4,5,6]**, compute its Euclidean norm manually and using NumPy.

Hint: By "manually," I mean solving it step by step using only basic NumPy functions.

### Answer 5

````python
a = np.array([1, 2, 3, 4, 5, 6])

# Manual: square -> sum -> square root
squared = a ** 2
sum_sq = np.sum(squared)
norm_manual = np.sqrt(sum_sq)
print("Squared components:", squared)
print("Sum of squares:    ", sum_sq)
print("Manual norm:       ", norm_manual)

# Using NumPy
norm_np = np.linalg.norm(a)
print("np.linalg.norm:    ", norm_np)
print("Equal:", np.isclose(norm_manual, norm_np))
````

## Q6. Generate a random vector of length 10 and normalize it. Print the norm before and after normalization

### Answer 6

````python
np.random.seed(42)
r = np.random.rand(10)
print("Random vector:", r)

norm_before = np.linalg.norm(r)
r_normalized = r / norm_before
norm_after = np.linalg.norm(r_normalized)

print("Norm before normalization:", norm_before)
print("Norm after normalization: ", norm_after)
print("Normalized vector:", r_normalized)
````

## Q7. Compute the dot product of vectors

a = [1,2,3,4,5,6] & b = [0,-4,3,-3,6,5]

using:

1. Manual loop
2. NumPy `np.dot()` .

### Answer 7

````python
a = [1, 2, 3, 4, 5, 6]
b = [0, -4, 3, -3, 6, 5]

# 1. Manual loop
dot_loop = 0
for x, y in zip(a, b):
    dot_loop += x * y
print("Manual loop:", dot_loop)

# 2. NumPy
dot_np = np.dot(np.array(a), np.array(b))
print("np.dot():   ", dot_np)
````

## Q8. Verify the distributive property:

$$a · (b + c) = a · b + a · c$$

Hint: Generate three random numbers

### Answer 8

````python
np.random.seed(0)
a = np.random.rand(5)
b = np.random.rand(5)
c = np.random.rand(5)

lhs = np.dot(a, b + c)
rhs = np.dot(a, b) + np.dot(a, c)

print("a · (b + c)     =", lhs)
print("a · b + a · c   =", rhs)
print("Property holds: ", np.isclose(lhs, rhs))
````

## Q9: Compute the Euclidean distance matrix between every point in group A and every point in group B.

```python
# Group A
A = np.array([
    [ 2,  1],
    [ 5,  4],
    [ 1,  6],
    [ 7,  2]
])

# Group B
B = np.array([
    [0, 0],
    [6, 3],
    [3, 5]
])
```

````python
# Group A
A = np.array([
    [ 2,  1],
    [ 5,  4],
    [ 1,  6],
    [ 7,  2]
])

# Group B
B = np.array([
    [0, 0],
    [6, 3],
    [3, 5]
])


diff = A[:, np.newaxis, :] - B[np.newaxis, :, :]
dist = np.sqrt(np.sum(diff ** 2, axis=2))   # shape (4, 3)

print("Shape:", dist.shape)
print(np.round(dist, 3))
````

````python
# Verification with an explicit double loop
dist_loop = np.zeros((len(A), len(B)))
for i in range(len(A)):
    for j in range(len(B)):
        dist_loop[i, j] = np.linalg.norm(A[i] - B[j])

print(np.allclose(dist, dist_loop))
````
