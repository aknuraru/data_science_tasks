## Q1: What is the Probability that the Customer has a Credit Card ?

```python
total=len(df) # number of total customers
p_card=(df['CreditCard']=='yes').sum()/total
p_card
```

```python
p_age50=(df['age']>50).sum()/total
p_age50
```

#### We need to use Bays Therorm to calculate the probability of having a rainfall more than 500 mm in month of June, provided it has flooded in that year ?

#### Follow the following instructions to answer this question

## 1- We may start by preparing the required columns:

1. Create a new column named "June_more_than_500" and assign a value of 1 if the amount is more than 500 and 0 if it is less
than 500.
2. Do the same for the 'FLOODS' column, assigning 1 for 'yes' and 0 for 'no'.

```python
df['June_more_than_500'] = (df['JUN'] > 500).astype('int')
df['FLOODS'] = df['FLOODS'].map({'YES': 1, 'NO': 0})
```

## 2- Select only the columns that will help to answer the main question.

```python
df2 = df[['YEAR', 'June_more_than_500', 'FLOODS']]
df2.head()
```

## 3- Create a cross tabulation for 'FLOODS' and 'June_more_than_500'

```python
ct = pd.crosstab(df2['FLOODS'], df2['June_more_than_500'],margins=True)
ct
```

## 4- What is the probability of a rainfall more than 500mm in June?

```python
p_A = 93/118
p_A
```

## 5- What is the probability of having a flood?

```python
p_B = 60/118
p_B
```

## 6- What is the probability of having a flood and a rainfall more than 500mm in June?

```python
p_B_AND_A = 54/118
p_B_AND_A
```

## 7- What is the Probability of having a flood given it rained more than 500mm in June?

```python
p_B_given_A = p_B_AND_A / p_A
p_B_given_A
```

## 8- What is the probability of a rainfall less than 500mm in June?

```python
p_not_A = 25/118
p_not_A
```

## 9- What is probability of having a flood given that it rained less than 500mm?

```python
p_B_given_not_A = 6/25
p_B_given_not_A
```

#### Now, you can answer the main question using Bays theorem.
#### Calculate the probability of having a rainfall more than 500mm in month of June, provided it has flooded in that year ?

## 10- Answer this question by using Bayes Theorem and the law of the Total Probaility

```python
p_A_given_B = (p_B_given_A * p_A) / ((p_A * p_B_given_A) + (p_not_A * p_B_given_not_A))
p_A_given_B
```

## 11- Answer this question by using Bayes Theorem only.

```python
p_A_given_B = p_B_given_A * p_A / p_B
p_A_given_B
```
