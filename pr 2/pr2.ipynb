### 11. How many products cost more than $10.00 ?

### 13.Sort the dataset by item name

```python
df.sort_values(by="item_name")
```

### 14.what is quantity of the most expensive item ordered?

```python
df.sort_values(by="item_price",ascending = False)["quantity"].iloc[0]
```

```python
df.sort_values(by="item_price",ascending = False)["quantity"].head(1).values[0]
```

### 15.How many times was a Veggie salad bowl ordered?

### 16. How many times did someone ordered more than 1 canned soda?

```python
condition = (df.item_name=="Canned Soda")&(df.quantity>1)
```
