```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# List
list=[1,2,3]
list
```

```python
#1d array
ar=np.array([1,2,3])
ar
```

```python
# row
row=np.array([{1,2,3}])
row
```

```python
# col
col=np.array([[1],[2],[3]])
col
```

### Orientation and shape

```python
print(f'List:{np.shape(list)}')
print(f'Array:{np.shape(ar)}')
print(f'Row:{np.shape(row)}')
print(f'Col:{np.shape(col)}')
```

### Geometric of vectors

```python
v1 = [5,-3]
# plot in 2D
plt.plot([0,v1[0]],[0,v1[1]])
plt.axis((-6,6, -6,6))
plt.plot([-6,6],[0,0], 'k--')
plt.plot([0,0],[-6,6], 'k--')
```

```python
# 3D dimension vector
v3=[5,-4,3]
ax=plt.axes(projection='3d')
ax.plot([0,v3[0]],[0,v3[1]],[0,v3[2]])
ax.plot([-5,5],[0,0],[0,0],'k--')
ax.plot([0,0],[-5,5],[0,0],'k--')
ax.plot([0,0],[0,0],[-6,6],'k--');
```

### Vector addiction and substraction

```python
a=np.array([[4,5,6]])
b=np.array([[10,20,30]])
c=np.array([[0,3,6,9]])
```

```python
a+b
```

```python
a+c
```

### Geometric of vectors additiion

```python
# 
v1= np.array([4,-1])
v2= np.array([3,5])
# Result
v3=v1+v2
v3
```

```python
plt.plot([0,v1[0]],[0,v1[1]],'b',label='v1')
plt.plot([0,v2[0]]+v1[0],[0,v2[1]]+v1[1],'r',label='v2')
plt.plot([0,v3[0]],[0,v3[1]],'g',label='v1+v2')
plt.plot([-7,7],[0,0],'k--')
plt.plot([0,0],[-7,7],'k--')

plt.legend()
```
