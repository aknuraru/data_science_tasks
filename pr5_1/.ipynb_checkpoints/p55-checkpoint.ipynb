# Question 1:

#### Create the following objects in NumPy:

- `List = [3, 4, 5]`
- `Ara = np.array([3, 4, 5])`
- `row = np.array([[3, 4, 5]])`
- `col = np.array([[3],[4],[5]])`

#### Then print their shapes exactly in the format shown below:

##### List: (?)
##### Array: (?)
##### Row: (?)
##### Column: (?)
##### (Replace `?` with the correct shape outputs.)

````python
import numpy as np

List = [3, 4, 5]
Ara = np.array([3, 4, 5])
row = np.array([[3, 4, 5]])
col = np.array([[3], [4], [5]])

print("List:", np.shape(List))
print("Array:", Ara.shape)
print("Row:", row.shape)
print("Column:", col.shape)
````

# Question 2:

#### Given:


```python
import numpy as np
a = np.array([[4,5,6]])
c = np.array([[0,3,6,9]])


#### **2.1. Compute `a + c` . What error do you get?**

````python
a = np.array([[4, 5, 6]])
c = np.array([[0, 3, 6, 9]])
a + c
````

#### **2.2. Explain why it happens using the idea of dimension and orientation compatibility.**

#### **Answer: a has shape (1, 3) and c has shape (1, 4). Both are row vectors, so they have the same orientation but different lengths (3 vs 4). Element-wise addition needs matching shapes, or dimensions where one of them is 1, so NumPy can't pair up the elements.**

#### **2.3. Add these two vectors without deleting or adding any elements.**

#### **Answer:Turn one vector into a column by transposing it. A (3,1) column plus a (1,4) row broadcasts into a (3,4) matrix.**

````python
result = a.T + c      # (3,1) + (1,4) -> (3,4)
print(result)
````

# Question 3:

#### Let the 2D vector be:

```python
v = np.array([4, 5])

### **Do the following in one plot:**

1. Plot the vector as an arrow or line segment from the origin `(0,0)` to `(5,-3)`.
2. Add dashed x- and y-axes similar to the one in the material provided (horizontal and
vertical dashed lines through the origin).
3. Set the plot limits so the vector is clearly visible (choose reasonable limits).

````python
import numpy as np
import matplotlib.pyplot as plt

v = np.array([5, -3])

plt.figure()
plt.plot([0, v[0]], [0, v[1]])
plt.axis((-6, 6, -6, 6))
plt.plot([-6, 6], [0, 0], 'k--')
plt.plot([0, 0], [-6, 6], 'k--')
plt.show()
````

# Question 4:

#### Let the 3D vector be:

```python
v3 = np.array([6, -5, 4])

### **Do the following in one plot:**

1. Create a 3D plot and draw the vector from the origin to `(5, -4, 3)`.
2. Label the axes as `x , y, z` .

````python
v3 = [5, -4, 3]   # question says draw to (5,-4,3)
plt.figure()
ax = plt.axes(projection='3d')
ax.plot([0, v3[0]], [0, v3[1]], [0, v3[2]])
ax.plot([-5, 5], [0, 0], [0, 0], 'k--')
ax.plot([0, 0], [-5, 5], [0, 0], 'k--')
ax.plot([0, 0], [0, 0], [-6, 6], 'k--')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.show()
````

# Question 5:

#### Given:

```python
v1 = np.array([4, -2])
v2 = np.array([3, 5])

#### **5.1. Compute `v3 = v1 + v2` and print `v3`**

````python
import numpy as np

v1 = np.array([4, -2])
v2 = np.array([3, 5])

v3 = v1 + v2
print(v3)
````

#### **5.2. On the same 2D figure, plot all three** vectors `v1` , `v2` , and `v3` from the origin.****

````python
# ---- Q5.2 ----
plt.figure()
plt.plot([0, v1[0]], [0, v1[1]], 'b', label='v1')
plt.plot([0, v2[0]], [0, v2[1]], 'r', label='v2')
plt.plot([0, v3[0]], [0, v3[1]], 'g', label='v1+v2')
plt.plot([-8, 8], [0, 0], 'k--')
plt.plot([0, 0], [-8, 8], 'k--')
plt.axis((-8, 8, -8, 8))
plt.legend()
plt.show()
````
